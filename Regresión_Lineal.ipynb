```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([50, 60, 70, 80, 90, 100, 110, 120])
y = np.array([150, 180, 195, 230, 260, 290, 325, 355])
```

```python
m = X.size
X_b = np.c_[np.ones((m, 1)), X]
```

```python
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print("Pendiente: ", round(theta[1], 2))
print("Intercepto: ", round(theta[0], 2))
```

```python
precio = theta[0] + theta[1] * 95
print("Precio estimado: ", round(precio, 2), "miles")
```

```python
plt.scatter(X, y, label='Datos')

m_cuadrados = np.array([X.min(), X.max()])
precio = theta[0] + theta[1] * m_cuadrados
plt.plot(m_cuadrados, precio, color='#7F2CCB', linestyle=':', label='Línea de Regresión')

plt.xlabel('Tamaño (m^2)')
plt.ylabel('Precio (MXN)')
plt.title('Precio de Vivienda por m^2')
plt.legend()
plt.grid(True)
plt.show()
```
